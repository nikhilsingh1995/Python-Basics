**5.**What is mutability in Python?

**Answer:** Mutability refers to the ability of an object to be changed after it has been created.
In Python, objects are either mutable or immutable depending on whether or not their values can be modified.

Mutable Objects:
Mutable objects can be changed after they are created.

This means that the data inside the object can be modified without changing the object's identity.

Examples:

Lists: We can change, add, or remove elements from a list.

Dictionaries: We can add, remove, or modify key-value pairs.

Sets: We can add or remove elements from a set.

Mutable objects are useful when we need to modify data in place or update data structures over time.

**6.**Why are lists mutable, but tuples are immutable?

**Answer:** Lists are mutable and tuples are immutable because they are designed for different purposes in python.Lists are mutable for flexibility, and tuples are immutable for safety, performance, and conceptual clarity.

Lists Are Mutable because of following:

**Flexibility**: Lists are meant to store collections of items that may need to be changed, expanded, or reduced during a program's execution.

**Use Case**: Ideal for scenarios where you need to add, remove, or modify elements frequently, such as managing dynamic datasets or accumulating results.

Tuples Are Immutable because of following:

**Data Integrity**: Tuples are used when the data shouldn't change after creation—this protects against accidental modification.

**Hashability**: Only immutable objects can be used as dictionary keys or elements in a set, and tuples can be used this way (if all their elements are also immutable).

**Performance**: Tuples are slightly faster than lists due to their immutability—Python can optimize memory and access speed.

**Semantic Meaning**: Tuples often represent a fixed collection of items (like a pair of coordinates or a database record) where changing individual elements doesn’t make sense.





**7**.What is the difference between “==” and “is” operators in Python ?

**Answer:**

 **== (Equality Operator):**

Purpose: Checks whether the values of two objects are equal.

Behavior: It compares the contents of the objects, regardless of whether they are stored at the same location in memory.

Usage: Best used when we care about whether two objects represent the same data.

**is (Identity Operator):**

Purpose: Checks whether two references point to the same object in memory.

Behavior: It compares the identity of the objects, not their contents.

Usage: Best used when we want to know if two variables refer to the exact same object, often important in situations involving None, singletons, or object caching.

**10.** What is the difference between implicit and explicit type casting ?

**Answer:**

**Implicit Type Casting**:

Done automatically by Python.

Happens when Python safely converts one type to another during operations.

No code is needed from the programmer.

**Ex:**

a = 5       # int

b = 2.0     # float

c = a + b   # int + float → float (implicit conversion)

print(c)        # Output: 7.0

print(type(c))  # Output: <class 'float'>

**Explicit Type Casting**:

Done manually by the programmer.

We use casting functions like int(), float(), str() to convert types.

Use when we need to control the type conversion.

**Ex:**


x = "10"        # string

y = int(x)      # explicitly cast to int

print(y + 5)    # Output: 15

**13.** What is the difference between for and while loops ?

**Answer:**

The key differences are:

For Loop:

- Used for iterating over a sequence (list, tuple, string, etc.)
- Number of iterations is predetermined by the sequence length
- Better when you know the number of iterations in advance
- Cleaner syntax for sequence iteration

While Loop:

- Executes as long as a condition remains True
- Number of iterations depends on when the condition becomes False
- Better when you don't know how many iterations you'll need
- More suitable for checking conditions that change during execution

**14.** Describe a scenario where a while loop is more suitable than a for loop.

**Answer:**

A while loop is more suitable than a for loop when you don’t know how many times something needs to happen — for example, when you want to keep asking the user for valid input.

Let's assume a scenario for this with following example:

**Ex:**

**6.** Write a function to demonstrate the behavior of mutable and immutable arguments.

```python
#Answer
def modify_arguments(a_list, b_num):
    # Modifying the list (mutable)
    a_list.append(73)

    # Modifying the number (immutable)
    b_num += 15

    print("Inside the function:")
    print("List:", a_list)
    print("Number:", b_num)

# Original values
c_list = [1, 2, 3]
d_num = 500

# Call the function
modify_arguments(c_list, d_num)

# Values after the function call
print("\nOutside the function:")
print("List:", c_list)       # Changed!
print("Number:", d_num)   # Unchanged
```

List is *mutable*, so changes inside the function affect the original list.

Integer is *immutable*, so changes inside the function don’t affect the original number(d_num) — a new local variable is created.

**7.** Write a program that performs basic arithmetic operations on two user-input numbers.

```python
#Answer
# Basic Airthmetic operations are addition(+),subtraction(-),multiplication(*),division(/),floor division(//),modulus(%),exponentiation(**)

#Lets take 2 inputs from User
a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))

# Perform arithmetic operations
addition = a+ b
subtraction = a- b
multiplication = a* b
exponent = a** b

# Display results
print("\nResults of arithmetic operations:")
print("Addition: " + str(a) + " + " + str(b) + " = " + str(addition))
print("Subtraction: " + str(a) + " - " + str(b) + " = " + str(subtraction))
print("Multiplication: " + str(a) + " * " + str(b) + " = " + str(multiplication))
print("Exponentiation: " + str(a) + " ** " + str(b) + " = " + str(exponent))

# Check for division by zero
if b != 0:
    division = a / b
    floor_div = a // b
    modulus = a % b
    print("Division: " + str(a) + " / " + str(b) + " = " + str(division))
    print("Floor Division: " + str(a) + " // " + str(b) + " = " + str(floor_div))
    print("Modulus: " + str(a) + " % " + str(b) + " = " + str(modulus))
else:
    print("Division, Floor Division, and Modulus are not allowed (cannot divide by zero).")
```

**9.** Write a Python program to convert user input from string to integer, float, and boolean types.

```python
# Answer
# input from user
x = input("Enter a value: ")
if x.isdigit():
    integer_value = int(x)
    print("Integer conversion:", integer_value)
else:
    print("The input cannot be converted to an integer.")

# Convert to float (check if the input can be converted to a float)
# We will try to split the string on '.' and check if both parts are digits
if x.replace('.', '', 1).isdigit() and x.count('.') <= 1:
    float_value = float(x)
    print("Float conversion:", float_value)
else:
    print("The input cannot be converted to a float.")

# Convert to boolean
# Any non-empty string is considered True, an empty string is False
boolean_value = bool(x)
print("Boolean conversion:", boolean_value)




```

**14.** Write a program to reverse a string using a while loop.

```python
# Answer

# input from user
word = input("Enter a word: ")

# Empty string to store the reverse
reverse = ""

# Set the position to the last letter in the word
i = len(word) - 1

# Use a while loop to go backward through the word
while i >= 0:
    reverse = reverse + word[i]
    i = i - 1

# Output of reversed word
print("Reversed word is:", reverse)
```

**15.** Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
#Answer

# Enter a number
a = int(input("Enter a number to find its factorial: "))

# Set the result
factorial = 1

# Make a copy of the number for looping
n = a

# while loop to calculate factorial
while n > 1:
    factorial = factorial * n
    n = n - 1

# Print the output
print("The factorial of", a, "is", factorial)
```
